```python
# Creating variables with different data types
market_name = "Gobachop Market"  # String data type
total_market_table = 150  # Integer data type
price_per_bag = 3200.50  # Float data type
is_market_open = True  # Boolean data type

# Printing variables with clear labels for the client
print("Market Name:", market_name)
print("Total Market Tables Available:", total_market_table)
print("Price Per Bag:", price_per_bag, "LRD")
print("Is the Market Open Today?:", is_market_open)

# Bonus: Printing the technical data types
print("Data type of market_name:", type(market_name))
print("Data type of total_market_table:", type(total_market_table))
print("Data type of price_per_bag:", type(price_per_bag))
print("Data type of is_market_open:", type(is_market_open))
```

```python
today_price = 3200  # Rice price today in LRD
increase = 150  # Price increase amount for tomorrow

tomorrow_price = today_price + increase  # Add the increase to today's price

print("Tomorrow's price for a bag of rice will be:", tomorrow_price, "LRD")
```

```python
total_onion_cost = 1800  # Total cost of a bag of onions in LRD
number_of_traders = 3  # Shared equally by 3 people

cost_per_trader = total_onion_cost / number_of_traders  # Divide total cost by number of traders

print("The cost split per trader is:", cost_per_trader, "LRD")
```

```python
original_price = 2500  # Original price of palm oil in LRD
discount_rate = 0.10  # 10% discount rate represented as a decimal

discount_amount = original_price * discount_rate  # Calculate the discount savings
final_price = original_price - discount_amount  # Subtract savings from the original price

print("Discount Amount Saved:", discount_amount, "LRD")
print("Final Price to Pay after Discount:", final_price, "LRD")
```
